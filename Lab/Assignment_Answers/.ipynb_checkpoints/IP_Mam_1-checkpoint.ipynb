Exp 3: Add Border (Padding)

```python
# Binary image with border (auto-convert if needed)
import cv2
import numpy as np

# Step 1: Read the image in grayscale
img = cv2.imread('/content/gray.webp', cv2.IMREAD_GRAYSCALE)

# Step 2: Check if the image is already binary
unique_values = set(img.flatten())
if not unique_values.issubset({0, 255}):
    print("Image is not binary.")
    print(f"Unique pixel values found: {sorted(unique_values)}")

    # Step 3: Convert to binary using thresholding
    _, img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
    print("Image has been converted to binary using thresholding at 127.")

else:
    print("Image is already binary.")

# Step 4: Get user input for border settings
border_width = int(input("Enter border width in pixels: "))
color = int(input("Enter border color (0 for black, 255 for white): "))  # Must be 0 or 255

# Step 5: Add border
bordered = cv2.copyMakeBorder(
    img,
    border_width, border_width,
    border_width, border_width,
    cv2.BORDER_CONSTANT,
    value=color
)

# Step 6: Save the output
cv2.imwrite('binary_bordered.jpg', bordered)
print("Bordered binary image saved as 'binary_bordered.jpg'")
```

using all available border styles

```python
import cv2

# Read color image
img = cv2.imread('/content/colored.jpg')

if img is None:
    raise FileNotFoundError("Image not found. Please check the path.")

# Ask for border width
border_width = int(input("Enter border width in pixels: "))
# Ask for constant color value for BORDER_CONSTANT (applies same value to all 3 channels)
constant_color = int(input("Enter border color (0–255) for constant border: "))

# Dictionary of border types and their OpenCV codes
border_types = {
    "BORDER_CONSTANT": cv2.BORDER_CONSTANT,
    "BORDER_REPLICATE": cv2.BORDER_REPLICATE,
    "BORDER_REFLECT": cv2.BORDER_REFLECT,
    "BORDER_WRAP": cv2.BORDER_WRAP,
    "BORDER_REFLECT_101": cv2.BORDER_REFLECT_101,
    "BORDER_TRANSPARENT": cv2.BORDER_TRANSPARENT
}

# Apply each border type
for name, border_type in border_types.items():
    # Set value only for BORDER_CONSTANT, others ignore 'value'
    if border_type == cv2.BORDER_CONSTANT:
        bordered = cv2.copyMakeBorder(
            img,
            border_width, border_width,
            border_width, border_width,
            border_type,
            value=(constant_color, constant_color, constant_color)
        )
    else:
        bordered = cv2.copyMakeBorder(
            img,
            border_width, border_width,
            border_width, border_width,
            border_type
        )

    # Save the result
    filename = f'{name.lower()}.jpg'
    cv2.imwrite(filename, bordered)
    print(f"Saved: {filename}")
```

```python
import cv2
import numpy as np

# Read the image
image = cv2.imread(r"D:\PhDWork\Image processing\f.jpg")

# Brightness adjustment factor
brightness_factor = 1.5  # Change this value to adjust brightness

# Adjust brightness using pixel-wise multiplication or division
if brightness_factor > 1:
    # Brighten the image by multiplying pixel values
    brightened_image = cv2.multiply(image, np.array([brightness_factor]))
else:
    # Darken the image by dividing pixel values
    darkened_image = cv2.divide(image, np.array([1 / abs(brightness_factor)]))

# Save the adjusted image
if brightness_factor > 1:
    cv2.imwrite('brightened_image1.jpg', brightened_image)
else:
    cv2.imwrite('darkened_image1.jpg', darkened_image)
```

# Tranformation

```python
# Import the necessary Libraries
import cv2
import matplotlib.pyplot as plt

# Read image from disk.
img= cv2.imread(r"D:\PhDWork\Image processing\panda2.jpg")


# Convert BGR image to RGB
image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Image rotation parameter
center = (image_rgb.shape[1] // 2, image_rgb.shape[0] // 2)
angle = -30
scale = 1

# getRotationMatrix2D creates a matrix needed for transformation.
rotation_matrix = cv2.getRotationMatrix2D(center, angle, scale)

# We want matrix for rotation w.r.t center to 30 degree without scaling.
rotated_image = cv2.warpAffine(image_rgb, rotation_matrix, (image.shape[1], image.shape[0]))

# Create subplots
fig, axs = plt.subplots(1, 2, figsize=(7, 4))

# Plot the original image
axs[0].imshow(image_rgb)
axs[0].set_title('Original Image')

# Plot the Rotated image
axs[1].imshow(rotated_image)
axs[1].set_title('Image Rotation')

# Remove ticks from the subplots
for ax in axs:
	ax.set_xticks([])
	ax.set_yticks([])

# Display the subplots
plt.tight_layout()
plt.show()
```

```python
# Import the necessary Libraries
import cv2
import matplotlib.pyplot as plt

# Read image from disk.
image = cv2.imread(r"D:\PhDWork\Image processing\panda2.jpg")

# Convert BGR image to RGB
image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

width = image_rgb.shape[1]
height = image_rgb.shape[0]

tx = -100
ty = 70

# Translation matrix
translation_matrix = np.array([[1, 0, tx], [0, 1, ty]], dtype=np.float32)
# warpAffine does appropriate shifting given the Translation matrix.
translated_image = cv2.warpAffine(image_rgb, translation_matrix, (width, height))

# Create subplots
fig, axs = plt.subplots(1, 2, figsize=(7, 4))

# Plot the original image
axs[0].imshow(image_rgb)
axs[0].set_title('Original Image')

# Plot the transalted image
axs[1].imshow(translated_image)
axs[1].set_title('Image Translation')

# Remove ticks from the subplots
for ax in axs:
	ax.set_xticks([])
	ax.set_yticks([])

# Display the subplots
plt.tight_layout()
plt.show()
```

```python
# Import the necessary Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
image = cv2.imread(r"D:\PhDWork\Image processing\panda2.jpg")


# Convert BGR image to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Image shape along X and Y
width = image_rgb.shape[1]
height = image_rgb.shape[0]

# Define the Shearing factor
shearX = -0.15
shearY = 0

# Define the Transformation matrix for shearing
transformation_matrix = np.array([[1, shearX, 0], [0, 1, shearY]], dtype=np.float32)
# Apply shearing
sheared_image = cv2.warpAffine(image_rgb, transformation_matrix, (width, height))

# Create subplots
fig, axs = plt.subplots(1, 2, figsize=(7, 4))

# Plot the original image
axs[0].imshow(image_rgb)
axs[0].set_title('Original Image')

# Plot the Sheared image
axs[1].imshow(sheared_image)
axs[1].set_title('Sheared image')

# Remove ticks from the subplots
for ax in axs:
	ax.set_xticks([])
	ax.set_yticks([])

# Display the subplots
plt.tight_layout()
plt.show()
```

```python

# Perform linear contrast stretching

import cv2
import numpy as np

# Read the image
image = cv2.imread(r"D:\PhDWork\Image processing\gettyimage.jpg", cv2.IMREAD_GRAYSCALE)

min_intensity = np.min(image)
max_intensity = np.max(image)
output_image = cv2.convertScaleAbs(image, alpha=255.0 / (max_intensity - min_intensity), beta=-min_intensity)
cv2.imshow("output_image ", output_image )
cv2.waitKey(0)
cv2.destroyAllWindows()
# Save the adjusted image
cv2.imwrite('output_image.jpg', output_image)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def generate_impulse_noise(image, probability):
    noisy_image = image.copy()
    total_pixels = image.shape[0] * image.shape[1]
    num_noisy_pixels = int(total_pixels * probability)

    # Generate random coordinates for noisy pixels
    noisy_pixels = np.random.randint(0, image.shape[0], size=(num_noisy_pixels,))
    noisy_rows = np.random.randint(0, image.shape[1], size=(num_noisy_pixels,))

    # Set noisy pixels to either black or white
    for i in range(num_noisy_pixels):
        x, y = noisy_pixels[i], noisy_rows[i]
        if np.random.rand() < 0.5:
            noisy_image[x, y] = 0  # Set to black
        else:
            noisy_image[x, y] = 255  # Set to white

    return noisy_image

img = cv2.imread('car.png')

# Generate impulse noise with a probability of 0.05 (adjust as needed)
impulse_noisy_img = generate_impulse_noise(img, probability=0.05)

# Display the images
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(impulse_noisy_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Impulse Noisy Image')

plt.show()
```

# DENOISING

```python


import cv2
import numpy as np
from skimage import img_as_ubyte

## original image
img = cv2.imread('f.jpg', 0)
img = img/img.max() # normalize the pixel value (0~1)


# Salt and Pepper Noise
x,y = img.shape
g = np.zeros((x,y), dtype=np.float32)
pepper = 0.1
salt = 0.95
for i in range(x):
    for j in range(y):
        rdn = np.random.random()
        if rdn < pepper:
            g[i][j] = 0
        elif rdn > salt:
            g[i][j] = 1
        else:
            g[i][j] = img[i][j]

img_noise = g

# preview the images
cv2.imshow('Original Image', img)
cv2.imshow('Image + Noise', img_noise)

cv2.waitKey(0)
cv2.destroyAllWindows()

## denoise image
# mean filter (average)
m = 5
n = 5
denoise_mean = cv2.blur(img_noise, (m,n))

# median filter
img_noise_median = np.clip(img_noise, -1, 1) #pixel value range
img_noise_median = img_as_ubyte(img_noise_median) #convert to uint8
denoise_median = cv2.medianBlur(img_noise_median, 5)

# preview the images
cv2.imshow('Original Image', img)
cv2.imshow('Image + Noise', img_noise)
cv2.imshow('Denoise Mean', denoise_mean)
cv2.imshow('Denoise Median', denoise_median)

cv2.waitKey(0)
cv2.destroyAllWindows()

# (optional) save the result
cv2.imwrite('Denoise mean.jpg', img_as_ubyte(denoise_mean))
cv2.imwrite('Denoise median.jpg', img_as_ubyte(denoise_median))
```

# Image smothening technqiues

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# original image
original_image = cv2.imread(r"D:\PhDWork\Image processing\noisy_img.jpg", 0)

# blur and sharpen convolution kernel
M = 3
blur_kernel = np.ones((M,M)) * 1/(M*M)

sharpen_kernel = np.array([[0, -1, 0],
                           [-1, 5, -1],
                           [0, -1, 0]],
                          dtype=np.float32)

# apply the covolution
blur_image = cv2.filter2D(src=original_image,
                          ddepth=-1,
                          kernel=blur_kernel)

sharpen_image = cv2.filter2D(src=original_image,
                             ddepth=-1,
                             kernel=sharpen_kernel)

# display the result
plt.imshow(original_image, cmap='gray')
plt.axis('off')
plt.show()

plt.imshow(blur_image, cmap='gray')
plt.axis('off')
plt.show()

plt.imshow(sharpen_image, cmap='gray')
plt.axis('off')
plt.show()

# built in image bluring
builtin_blur = cv2.blur(src=original_image, ksize=(3,3))

plt.imshow(builtin_blur, cmap='gray')
plt.axis('off')
plt.show()

builtin_median = cv2.medianBlur(src=original_image, ksize=3)

plt.imshow(builtin_median, cmap='gray')
plt.axis('off')
plt.show()

builtin_gaussian = cv2.GaussianBlur(src=original_image,
                                    ksize=(3,3),
                                    sigmaX=0,
                                    sigmaY=0)

plt.imshow(builtin_gaussian, cmap='gray')
plt.axis('off')
plt.show()
```

# image sharpening using sobel operator



```python
import cv2
import numpy as np

# Load the image in grayscale mode
image = cv2.imread(r"D:\PhDWork\Image processing\f.jpg", 0)  # Already in grayscale

# Apply Sobel operators to calculate gradients
gradient_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)
gradient_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)

# Convert the gradients to the same data type as the input image
gradient_x = cv2.convertScaleAbs(gradient_x)
gradient_y = cv2.convertScaleAbs(gradient_y)

# Combine gradients to obtain the sharpened image
sharpened_image = cv2.addWeighted(image, 1.5, gradient_x, -0.5, 0)

# Display the original and sharpened images
cv2.imshow("Original Image", image)
cv2.imshow("Sharpened Image", sharpened_image)

# Wait for a key press and close the image windows
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# USING SOBEL OPERATOR

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image in grayscale
image = cv2.imread(r"D:\PhDWork\Image processing\f.jpg", 0)

# Apply Sobel operators to calculate gradients
gradient_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)
gradient_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)

# Calculate the gradient magnitude
gradient_magnitude = np.sqrt(gradient_x**2 + gradient_y**2)

# Normalize the gradient magnitude to the range [0, 255]
gradient_magnitude = cv2.normalize(gradient_magnitude, None, 0, 255, cv2.NORM_MINMAX, cv2.CV_8U)

# Display the original image and the gradient magnitude
plt.figure(figsize=(12, 6))
plt.subplot(121), plt.imshow(image, cmap='gray')
plt.title('Original Image'), plt.axis('off')
plt.subplot(122), plt.imshow(gradient_magnitude, cmap='gray')
plt.title('Sobel Edge Detection'), plt.axis('off')
plt.show()
```

# USING ROBERTS

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image in grayscale
image = cv2.imread(r"D:\PhDWork\Image processing\f.jpg", 0)
# Define the Roberts Cross kernels
roberts_x = np.array([[1, 0], [0, -1]])
roberts_y = np.array([[0, 1], [-1, 0]])

# Apply Roberts Cross operators
gradient_x = cv2.filter2D(image, -1, roberts_x)
gradient_y = cv2.filter2D(image, -1, roberts_y)

# Calculate the gradient magnitude
gradient_magnitude = np.sqrt(gradient_x**2 + gradient_y**2)

# Normalize the gradient magnitude to the range [0, 255]
gradient_magnitude = ((gradient_magnitude - gradient_magnitude.min()) * 255 / (gradient_magnitude.max() - gradient_magnitude.min())).astype(np.uint8)

# Display the original image and the edge detection result
plt.figure(figsize=(12, 6))
plt.subplot(121), plt.imshow(image, cmap='gray')
plt.title ('Original Image'), plt.axis('off')
plt.subplot(122), plt.imshow(gradient_magnitude, cmap='gray')
plt.title('Roberts Cross Edge Detection'), plt.axis('off')
plt.show()
```

# PREWITT OPERATOR

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image in grayscale
image = cv2.imread(r"D:\PhDWork\Image processing\f.jpg", 0)

# Define the Prewitt kernels
prewitt_x = np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]])
prewitt_y = np.array([[-1, -1, -1], [0, 0, 0], [1, 1, 1]])

# Apply Prewitt operators
gradient_x = cv2.filter2D(image, -1, prewitt_x)
gradient_y = cv2.filter2D(image, -1, prewitt_y)

# Calculate the gradient magnitude
gradient_magnitude = np.sqrt(gradient_x**2 + gradient_y**2)

# Normalize the gradient magnitude to the range [0, 255]
gradient_magnitude = ((gradient_magnitude - gradient_magnitude.min()) * 255 / (gradient_magnitude.max() - gradient_magnitude.min())).astype(np.uint8)

# Display the original image and the edge detection result
plt.figure(figsize=(12, 6))
plt.subplot(121), plt.imshow(image, cmap='gray')
plt.title('Original Image'), plt.axis('off')
plt.subplot(122), plt.imshow(gradient_magnitude, cmap='gray')
plt.title('Prewitt Edge Detection'), plt.axis('off')
plt.show()
```

# Morphologocal operation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load an image (you should replace 'image.jpg' with your image file)
image = cv2.imread(r"D:\PhDWork\Image processing\coin1.jpg", 0)

# Define a kernel for morphological operations
kernel = np.ones((5, 5), np.uint8)

# Erosion
erosion = cv2.erode(image, kernel, iterations=1)

# Dilation
dilation = cv2.dilate(image, kernel, iterations=1)

# Opening (Erosion followed by Dilation)
opening = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel)

# Closing (Dilation followed by Erosion)
closing = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel)

# Display the original image and results
plt.figure(figsize=(15, 8))

plt.subplot(231), plt.imshow(image, cmap='gray')
plt.title('Original Image'), plt.axis('off')

plt.subplot(232), plt.imshow(erosion, cmap='gray')
plt.title('Erosion'), plt.axis('off')

plt.subplot(233), plt.imshow(dilation, cmap='gray')
plt.title('Dilation'), plt.axis('off')

plt.subplot(234), plt.imshow(opening, cmap='gray')
plt.title('Opening'), plt.axis('off')

plt.subplot(235), plt.imshow(closing, cmap='gray')
plt.title('Closing'), plt.axis('off')

plt.show()
```
