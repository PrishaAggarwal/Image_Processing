LOG TRANSFORM <br>
ismein yeh overall image ko bhi consider krta hai....ek image mein max pixel kya hai oske basis par baaki pixel value ko changes <br>
C log base 10 (1+pixel value) = new pixel value ....for each pixel value <br>
C = L/log base 10 (1+L) where L is maximum pixel value in particular image<br>
g(1,1) = new pixel value at given position <br>
f(1,1) =  old pixel value at given position <br>
Let C =1 <br>
then g(1,1) = 1*log base 10 (f(1,1)) <br> 
g(1,1) = log base 10 (129) = 2.11 (let 129 be pixel at that position) <br>
now we have tranformed values <br>
