```python
import cv2
image = cv2.imread("/content/image.webp") #in brackets its path of image
```

```python
from google.colab.patches import cv2_imshow #(for colab)
h,w = image.shape[:2] #it has 3 things, height,width and number of channels
# blue is one channel, green is one channel, red is one channel
#channel means rgb wali intensity but we only want height and width here
print("Height={}, Width={}".format(h,w))
cv2_imshow(image) #imshow to show display image and "" ke andar title of image and 2nd is image variable
```

```python
(B,G,R) = image[10,20] #rgb and bgr are 2 diff colours
print("R={},G={},B={}".format(R,G,B)) # .format because we want our result in rgb format
```

```python
B=image[100,100,0] #here last parameter is channel number i.e 0 and 0 stands for blue channel
print("B={}".format(B)) #we want only want one colour here i.e blue
```

```python
#converting grayscale image to bgr and vica versa we are using a function cvtcolor
gray_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY) # in 2nd parameter here we mentioning konsi image ko kismien convert krna hai
cv2.imshow("hehe",gray_image)
cv2.imwrite('test_write1.jpg',gray_image) #giving name to the new image we created and saving it
gray_image.shape #shape of gray scale image and it has only 2 parameters and no parameter like channel...since by defaault 1 channel i.e 0 to 255
```
