```python
#we now are applying colour map
#lets first check if our image is in grayscale or not using another method
gray_image = cv2.imread('test_write1.jpg',cv2.IMREAD_GRAYSCALE)
#now lets apply color map to our image
color_mapped = cv2.applyColorMap(gray_image,cv2.COLORMAP_JET) #color map has alot of colors under it like cool,jet etc that we have to specify

#display and save
cv2.imshow("DJ",color_mapped)
cv2.imwrite("hlo.jpg",color_mapped)
```

```python
#bgr to hsv
image1 = cv2.cvtColor(image,cv2.COLOR_BGR2HSV)
cv2.imshow("hdhd",image1)
cv2.imwrite('123.jpg',image1)
```

```python
#converting gray scale image to binary image
#in binary image we have only 2 values i.e 0 and 255
# we will assume mean intensity image as threshold value is one option
# 2nd option is getting threshold value from user
#threshold value is basically jo isse values kam hogi voh black mein convert ho jayega and isse bada wala white mein convert ho jayega
#numpy array works with matricxes so we are gonna use it now
# while reading image we can remove 2nd parameter, its just making sure that image is gray scale
import numpy as np
img = cv2.imread("C:/Users/csed/test_write1.jpg",cv2.IMREAD_GRAYSCALE)
#in threshold function we are giving image , threshold value , max value of gray scale, kismein convert krna i.e binary here)
mean_value = np.mean(img)
_, binary = cv2.threshold(img,mean_value,255,cv2.THRESH_BINARY) #gives 2 results i.e threshold value and image
#image here is stored in binary variable
#write image and save it
cv2.imwrite('binary_image.jpg',binary)
```
