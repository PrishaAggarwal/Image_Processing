```python
#convert gray scale to binary scale image
#case 1

import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Load grayscale image
gray_img = cv2.imread('/content/mountains.jpg', cv2.IMREAD_GRAYSCALE)

# Calculate mean intensity (threshold)
threshold = np.mean(gray_img)

# Convert to binary image
_, binary_img = cv2.threshold(gray_img, threshold, 255, cv2.THRESH_BINARY)

cv2.imwrite('binary_image_case1.jpg', binary_img)
# cv2_imshow('binary_image_case1.jpg', binary_img) for jupyter notebook
cv2_imshow(binary_img)
```

```python
#convert rgb to gray scale
#case1

color_img = cv2.imread('/content/rose.jpg')
gray_img = np.mean(color_img, axis=2).astype(np.uint8) #color_img is our image that has ht,wd,channel
#np.mean calculates mean accross 3 planes and axis=2 means take avg along 3rd dimension i.e channels here
#so avg of rgb values is taken for each pixel
# the result is 2d image with ht and wd where each pixel is avg of avg intensity of og pixels
#since np.mean returns float value so astype converts that float value back to 8-bit unassigned integers(0-255)

# exmaple (120+130+123)/3 = something
#this procudes a single grayscale value representing the brughtness of that pixel
cv2.imwrite('grayscale_mean.jpg', gray_img)
cv2_imshow(gray_img)
```

```python
#convert rgb to gray scale
#case2

R = float(input("Enter weightages for R (sum of RGB must be 1): "))
G = float(input("Enter weightages for G (sum of RGB must be 1): "))
B = float(input("Enter weightages for B (sum of RGB must be 1): "))

# Ensure sum is 1
if abs(R + G + B - 1) > 1e-6:
    print("Sum must be 1") #abs means absolute value
#if the sum is 1 then the grayscale values stays within 0-255 for 8-bit images

# 0-255 is is standard range for 8-bit image
#0-1 we sometimes use for normalized image to make things easy

# Convert to grayscale using weights
gray_img_weighted = (R * color_img[:, :, 2] + G * color_img[:, :, 1] + B * color_img[:, :, 0]).astype(np.uint8)
#the input from user is basically the weight of each channel
#in open cv we use bgr so tabhi channel ke number aise hia
#we multiply each channel by its weight and add them
#once added we convert them into 8 bit i.e 0-255
#so we weighted the channels acc to what we want

cv2.imwrite('grayscale_weighted.jpg', gray_img_weighted)
cv2_imshow(gray_img_weighted)
```

```python
#wap a program to draw border around input image

def add_border(image_path):
    img = cv2.imread(image_path)
    border_width = int(input("Enter border width (in pixels): "))
    b = int(input("Enter border color in B format (0-255 each): "))
    g = int(input("Enter border color in G format (0-255 each): "))
    r = int(input("Enter border color in R format (0-255 each): "))

    bordered_img = cv2.copyMakeBorder(img, border_width, border_width, border_width, border_width,
                                      cv2.BORDER_CONSTANT, value=[b, g, r]) #cv2.BORDER_CONSTANT fills the border with const value i.e border will be solid
    cv2.imwrite('bordered_image.jpg', bordered_img)
    cv2_imshow(bordered_img)

# Use for grayscale or binary images as well
add_border('/content/beach.jpg')
```
